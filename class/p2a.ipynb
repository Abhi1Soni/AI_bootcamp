```python
# INTRODUCTION TO NUMPY FOR AI

import numpy as np
import matplotlib.pyplot as plt

print("Welcome to NumPy - The Foundation of AI and Data Analysis!")

# STEP 1: CREATE A 2D RANDOM DATASET

x = np.random.randint(1, 100, size=(4000, 1000))
print(x)

# STEP 2: CREATE A CRICKET DATASET

players = np.array(['Sachin', 'Virat Kohli', 'Yashasvi', 'Jaspreet', 'Rishab'])
matches = np.array(['IPL 2015', 'IPL 2020', 'IPL 2023'])
runs = np.random.randint(10, 100, size=(len(players), len(matches)))

print("Players:", players)
print("Matches:", matches)
print("Runs scored by each player:")
print(runs)

# STEP 3: SUM OF RUNS

total_runs_by_player = runs.sum(axis=1)
print("Total runs scored by each player:")
for player, total in zip(players, total_runs_by_player):
    print(f"{player}: {total} runs")

# STEP 4: AVERAGE RUNS

average_runs_per_match = runs.mean(axis=0)
print("Average runs scored in each match:")
for match, avg in zip(matches, average_runs_per_match):
    print(f"{match}: {avg:.2f} runs")

# STEP 5: PLAYER WITH MOST RUNS

most_runs_index = np.argmax(total_runs_by_player)
print(f"The player with the most runs is {players[most_runs_index]} with {total_runs_by_player[most_runs_index]} runs.")

# STEP 6: STANDARD DEVIATION

std_dev_by_player = runs.std(axis=1)
print("Variability in performance:")
for player, std in zip(players, std_dev_by_player):
    print(f"{player}: {std:.2f}")

# STEP 7: VISUALIZATION

plt.figure(figsize=(8, 5))
plt.bar(players, total_runs_by_player)
plt.title("Total Runs Scored by Each Player")
plt.xlabel("Players")
plt.ylabel("Total Runs")
plt.show()

# STEP 8: OLYMPIC DATASET

countries = np.array(['USA', 'China', 'India', 'Germany', 'Australia'])
years = np.array([2012, 2016, 2020, 2024])
np.random.seed(42)
medals = np.random.randint(10, 100, size=(len(countries), len(years), 3))

print("Countries:", countries)
print("Years:", years)
print("Medals (Gold, Silver, Bronze):")
print(medals)

# STEP 9: TOTAL MEDALS BY COUNTRY

total_medals_by_country = medals.sum(axis=2)
print("Total medals by each country over the years:")
for country, total in zip(countries, total_medals_by_country):
    print(f"{country}: {total}")

# STEP 10: MOST MEDALS IN 2020

medals_2020 = total_medals_by_country[:, 2]
most_medals_index = np.argmax(medals_2020)
print(f"Country with the most medals in 2020: {countries[most_medals_index]} ({medals_2020[most_medals_index]} medals)")

# STEP 11: AVERAGE MEDALS BY YEAR

average_medals_by_year = total_medals_by_country.mean(axis=0)
print("Average medals won by all countries in each year:")
for year, avg in zip(years, average_medals_by_year):
    print(f"{year}: {avg:.2f}")

# STEP 12: STANDARD DEVIATION

std_dev_by_year = total_medals_by_country.std(axis=0)
print("Standard deviation of medals won by all countries in each year:")
for year, std in zip(years, std_dev_by_year):
    print(f"{year}: {std:.2f}")

# STEP 13: VISUALIZATION

plt.figure(figsize=(10, 6))
for i, country in enumerate(countries):
    plt.plot(years, total_medals_by_country[i], label=country, marker='o')
plt.title('Total Olympic Medals by Country Over Years')
plt.xlabel('Year')
plt.ylabel('Total Medals')
plt.legend()
plt.grid(True)
plt.show()
```
