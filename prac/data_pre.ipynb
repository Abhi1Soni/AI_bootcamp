```python
import pandas as pd
import numpy as np

data = {
    "ID": [101, 102, 103, 104, 105, 106, 107, 108,
           109, 110, 111, 112, 113, 114, 115, 115],

    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohit", "Anjali",
             "Karan", "Neha", "Arjun", "Pooja", "Vikash", "Riya",
             "Sameer", "Kavita", "Mohit", "Mohit"],

    "Age": [25, 28, np.nan, 24, 30, 27, 35, 26,
            29, np.nan, 31, 23, 28, 32, 27, 27],

    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Male"],

    "City": ["Delhi", "Mumbai", "Delhi", np.nan, "Bangalore", "Mumbai",
             "Delhi", "Bangalore", "Delhi", "Mumbai", np.nan, "Delhi",
             "Bangalore", "Mumbai", "Delhi", "Delhi"],

    "Salary": [45000, 55000, 48000, 40000, np.nan, 52000,
               75000, 50000, 60000, 58000, 65000, np.nan,
               54000, 70000, 51000, 51000],

    "Experience": [2, 4, 3, 1, 6, np.nan, 10, 3,
                   5, 4, 7, 1, 4, 8, np.nan, np.nan],

    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR",
                   np.nan, "Finance", "IT", "HR", "Finance", "IT",
                   "IT", "HR", "Finance", "Finance"],

    "Performance_Score": [8.5, 9.0, 7.5, 8.0, 9.2, 8.7,
                          9.5, np.nan, 8.8, 9.1, 7.8, 8.2,
                          8.6, 9.3, 7.9, 7.9]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 1. Check dataset
df.head()
df.info()
df.shape

# 2. Find null values
df.isnull().sum()
```

```python
# 3. Find duplicate rows
df.duplicated().sum()

# 4. Remove duplicates
df.drop_duplicates(inplace=True)
```

```python

# 5. Fill numerical null values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Experience"] = df["Experience"].fillna(df["Experience"].mean())
```

```python

# 6. Fill categorical null values
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])
```

```python
# 7. Fill Performance Score
df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].mean()
)
```

```python
# 8. Check again
print(df.isnull().sum())
```
