```python
import math


def calculator():
    while True:
        print("\n" + "=" * 45)
        print("           PYTHON SCIENTIFIC CALCULATOR")
        print("=" * 45)

        print("""
1.  Addition
2.  Subtraction
3.  Multiplication
4.  Division
5.  Power
6.  Square Root
7.  Percentage
8.  Factorial
9.  Sin
10. Cos
11. Tan
12. Logarithm (base 10)
13. Natural Logarithm (ln)
14. Exponential (e^x)
15. Absolute Value
16. Floor
17. Ceiling
18. Modulus
19. Exit
""")

        choice = input("Enter your choice (1-19): ")

        try:

            # Addition
            if choice == "1":
                a = float(input("Enter first number: "))
                b = float(input("Enter second number: "))
                print("Result:", a + b)

            # Subtraction
            elif choice == "2":
                a = float(input("Enter first number: "))
                b = float(input("Enter second number: "))
                print("Result:", a - b)

            # Multiplication
            elif choice == "3":
                a = float(input("Enter first number: "))
                b = float(input("Enter second number: "))
                print("Result:", a * b)

            # Division
            elif choice == "4":
                a = float(input("Enter first number: "))
                b = float(input("Enter second number: "))

                if b == 0:
                    print("Error: Cannot divide by zero.")
                else:
                    print("Result:", a / b)

            # Power
            elif choice == "5":
                base = float(input("Enter base: "))
                exponent = float(input("Enter exponent: "))
                print("Result:", base ** exponent)

            # Square root
            elif choice == "6":
                num = float(input("Enter number: "))

                if num < 0:
                    print("Error: Cannot calculate square root of a negative number.")
                else:
                    print("Result:", math.sqrt(num))

            # Percentage
            elif choice == "7":
                number = float(input("Enter number: "))
                percentage = float(input("Enter percentage: "))

                result = (number * percentage) / 100
                print(f"{percentage}% of {number} =", result)

            # Factorial
            elif choice == "8":
                num = int(input("Enter a non-negative integer: "))

                if num < 0:
                    print("Error: Factorial is not defined for negative numbers.")
                else:
                    print("Result:", math.factorial(num))

            # Sin
            elif choice == "9":
                angle = float(input("Enter angle in degrees: "))
                result = math.sin(math.radians(angle))
                print("sin(", angle, ") =", result)

            # Cos
            elif choice == "10":
                angle = float(input("Enter angle in degrees: "))
                result = math.cos(math.radians(angle))
                print("cos(", angle, ") =", result)

            # Tan
            elif choice == "11":
                angle = float(input("Enter angle in degrees: "))
                result = math.tan(math.radians(angle))
                print("tan(", angle, ") =", result)

            # Log base 10
            elif choice == "12":
                num = float(input("Enter positive number: "))

                if num <= 0:
                    print("Error: Logarithm requires a positive number.")
                else:
                    print("Result:", math.log10(num))

            # Natural logarithm
            elif choice == "13":
                num = float(input("Enter positive number: "))

                if num <= 0:
                    print("Error: ln requires a positive number.")
                else:
                    print("Result:", math.log(num))

            # Exponential
            elif choice == "14":
                num = float(input("Enter value of x: "))
                print("e^x =", math.exp(num))

            # Absolute value
            elif choice == "15":
                num = float(input("Enter number: "))
                print("Absolute value:", abs(num))

            # Floor
            elif choice == "16":
                num = float(input("Enter number: "))
                print("Floor:", math.floor(num))

            # Ceiling
            elif choice == "17":
                num = float(input("Enter number: "))
                print("Ceiling:", math.ceil(num))

            # Modulus
            elif choice == "18":
                a = float(input("Enter first number: "))
                b = float(input("Enter second number: "))

                if b == 0:
                    print("Error: Cannot calculate modulus with zero.")
                else:
                    print("Result:", a % b)

            # Exit
            elif choice == "19":
                print("\nThank you for using the calculator!")
                break

            else:
                print("Invalid choice. Please select 1-19.")

        except ValueError:
            print("Error: Please enter a valid number.")

        except OverflowError:
            print("Error: Number is too large.")


calculator()
```
