```python
import numpy as np
players = ['Player1', 'Player2', 'Player3', 'Player4']
matches = ['Match1', 'Match2', 'Match3', 'Match4']
np.random.seed(42)  # Set a random seed for reproducibility
runs = np.random.randint(10,100,size=(len(players),len(matches)))

print("players:", players)
print("matches:", matches)
print("runs:", runs)
```

```python
total_runs = np.sum(runs, axis=1)
print("Total runs scored by each player:", total_runs)
```

```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Neha", "Vikas"],
    "Maths": [85, 72, 90, 65, 78],
    "Science": [80, 75, 88, 70, 82],
    "English": [75, 80, 92, 68, 76],
    "Computer": [90, 85, 95, 72, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Select only numeric columns
marks = df[["Maths", "Science", "English", "Computer"]]

# Sum
print(marks.sum())

# Mean (average)
print(marks.mean())

# Median
print(marks.median())

# Mode
print(marks.mode())

# Minimum
print(marks.min())

# Maximum
print(marks.max())

# Standard deviation
print(marks.std())

# Variance
print(marks.var())

# Count
print(marks.count())

# Range
print(marks.max() - marks.min())

# All statistics together
print(marks.describe())
```

```python
# Add total marks for each student
df["Total"] = marks.sum(axis=1)

# Add average marks for each student
df["Average"] = marks.mean(axis=1)

print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

time= np.linspace(0,10,300)
amplitude = np.sin(2*np.pi*time)
```

```python
fig,ax = plt.subplots(figsize =(8,4.5), dpi =100)
ax.plot(time,amplitude, color ="crimson", linewidth =2, label ="Sensor A displacement")
plt.show()
```
