```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# cv2.imread() is a function in the OpenCV library that is used to read an image from a specified file path. It takes the file path as an argument and returns the image as a NumPy array. The image can be in various formats such as JPEG, PNG, BMP, etc.
img = cv2.imread(r"C:\Users\HP\OneDrive\Desktop\Abhi\img.webp")
print(type(img))
print(img.shape)
plt.imshow(img)
cv2.imshow("Image", img)
cv2.waitKey(0)


```

```python
img = cv2.imread(r"C:\Users\HP\OneDrive\Desktop\Abhi\img.webp")
img_resize= cv2.resize(img, (100, 156))
cv2.show("Resized Image", img_resize)
cv2.waitKey(0)
```

```python
img = cv2.imread(r"C:\Users\HP\OneDrive\Desktop\Abhi\img.webp")

img_flip= cv2.flip(img, 0)
cv2.imshow("Flipped Image", img_flip)
cv2.waitKey(0)
```

```python
img = cv2.imread(r"C:\Users\HP\OneDrive\Desktop\Abhi\img.webp")
img_crop= img[0:100, 0:100]
cv2.imshow("Cropped Image", img_crop)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
