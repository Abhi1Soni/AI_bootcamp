```python
# ============================================================
# AI BOOTCAMP - DATA PROCESSING & VISUALIZATION EXPERIMENTS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re


# ============================================================
# 1. CREATE DATASET
# ============================================================

data = {
    'Visitor_ID': [f'V{1000 + i}' for i in range(20)],

    'Name': [
        'ABHISHEK SANDEEP      ZADE',
        'ARNAV AJAY DESHPANDE..',
        '.. .. ASHWINI LALCHAND MUNDAWARE/',
        'GAYATRI SURESH GAIKWAD',
        'HARSHADA GANESH CHAUDHARI',
        'VAIBHAVI HARISHWAR PATIL',
        np.nan,
        'VISHAKHA PUNDLIK JADHAV',
        'YASH BHARAT SOLUNKE/',
        'VIVEK SANTOSH KHANDWE',
        'VISHAKHA PUNDLIK JADHAV',
        'Tanushree chhanwal',
        'Shruti jaiswal',
        'Shriyash Sulakhe',
        'YASH BHARAT SOLUNKE',
        np.nan,
        'ARNAV AJAY DESHPANDE',
        'RUTUJA SANTOSH THOTE',
        'ROHIT DILIP BILWAL',
        'RITESH SHIVAJI BAIRAGI'
    ],

    'Age': [
        25, 23, 22, np.nan, 21, 25, 24, 24, 28, np.nan,
        22, 23, 25, 27, 25, 30, 31, 26, 19, 19
    ],

    'Ticket_Price': [
        500, 750, 500, 1000, np.nan, 500, 700, 650, 750, 1000,
        500, 800, np.nan, 750, 500, 700, 900, 850, 750, np.nan
    ],

    'Check_In_Time': [
        '10:00 AM', '10:30 AM', '11:00 AM', '11:30 AM', np.nan,
        '10:00 AM', '12:00 PM', '12:30 PM', '01:00 PM', '01:30 PM',
        '11:00 AM', '02:00 PM', np.nan, '02:30 PM', '10:00 AM',
        '12:00 PM', '03:00 PM', '03:30 PM', '04:00 PM', np.nan
    ],

    'City': [
        'Delhi', 'Aurangabad', 'Mumbai', 'Bombay', 'New Delhi',
        'NDL', 'Chennai', 'Chenai', 'Chennaai', 'Bangalore',
        'Delhi', 'Pune', 'New Delhi', 'Bombay', 'New Delhi',
        'NDL', 'Indore', 'Bangalore', 'Ujjain', 'Bangalore'
    ],

    'State': [
        'Delhi', 'Maharastra', 'Maharastra', 'Maharastra', 'Delhi',
        'Delhi', 'Tamilnadu', 'Tamilnadu', 'Tamilnadu', 'Karnataka',
        'Delhi', 'Maharastra', 'Delhi', 'Maharastra', 'Delhi',
        'Delhi', 'Madhya Pradesh', 'Karnataka', 'Madhya Pradesh',
        'Karnataka'
    ]
}


# Convert dictionary into Pandas DataFrame
df = pd.DataFrame(data)


# ============================================================
# 2. DISPLAY ORIGINAL DATA
# ============================================================

print("\n" + "=" * 60)
print("ORIGINAL DATASET")
print("=" * 60)

print(df)


# ============================================================
# 3. SAVE DATASET AS CSV
# ============================================================

df.to_csv("visitor_data.csv", index=False)

print("\nCSV file created successfully: visitor_data.csv")


# ============================================================
# 4. BASIC PANDAS INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("\nShape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 records:")
print(df.head())


# ============================================================
# 5. FIND MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

print(df.isnull().sum())


# ============================================================
# 6. DATA CLEANING USING REGEX
# ============================================================

print("\n" + "=" * 60)
print("REGEX DATA CLEANING")
print("=" * 60)


# Convert Name column to string temporarily
df['Name'] = df['Name'].astype('string')


# Remove leading and trailing spaces
df['Name'] = df['Name'].str.strip()


# Remove dots and slashes
df['Name'] = df['Name'].str.replace(r'[./]', '', regex=True)


# Remove multiple spaces between words
df['Name'] = df['Name'].str.replace(r'\s+', ' ', regex=True)


# Convert names to title case
df['Name'] = df['Name'].str.title()


print("\nCleaned Names:")
print(df['Name'])


# ============================================================
# 7. CLEAN CITY NAMES
# ============================================================

print("\n" + "=" * 60)
print("CITY CLEANING")
print("=" * 60)

# Replace different names referring to the same city

city_mapping = {
    'Bombay': 'Mumbai',
    'New Delhi': 'Delhi',
    'NDL': 'Delhi',
    'Chenai': 'Chennai',
    'Chennaai': 'Chennai'
}

df['City'] = df['City'].replace(city_mapping)

print(df['City'].value_counts())


# ============================================================
# 8. CLEAN STATE NAMES
# ============================================================

state_mapping = {
    'Maharastra': 'Maharashtra',
    'Tamilnadu': 'Tamil Nadu'
}

df['State'] = df['State'].replace(state_mapping)

print("\nCleaned States:")
print(df['State'].value_counts())


# ============================================================
# 9. HANDLE MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("HANDLING MISSING VALUES")
print("=" * 60)


# Fill Age with median
df['Age'] = df['Age'].fillna(df['Age'].median())


# Fill Ticket Price with median
df['Ticket_Price'] = df['Ticket_Price'].fillna(
    df['Ticket_Price'].median()
)


# Fill missing Check-In-Time with mode
df['Check_In_Time'] = df['Check_In_Time'].fillna(
    df['Check_In_Time'].mode()[0]
)


# Fill missing Name
df['Name'] = df['Name'].fillna('Unknown')


print("\nMissing values after cleaning:")
print(df.isnull().sum())


# ============================================================
# 10. STATISTICAL CONCEPTS
#     MEAN, MEDIAN, MODE, STANDARD DEVIATION
# ============================================================

print("\n" + "=" * 60)
print("STATISTICAL ANALYSIS")
print("=" * 60)


# AGE STATISTICS

print("\nAGE STATISTICS")

print("Mean Age:",
      df['Age'].mean())

print("Median Age:",
      df['Age'].median())

print("Mode Age:",
      df['Age'].mode()[0])

print("Standard Deviation of Age:",
      df['Age'].std())


# TICKET PRICE STATISTICS

print("\nTICKET PRICE STATISTICS")

print("Mean Ticket Price:",
      df['Ticket_Price'].mean())

print("Median Ticket Price:",
      df['Ticket_Price'].median())

print("Mode Ticket Price:",
      df['Ticket_Price'].mode()[0])

print("Standard Deviation of Ticket Price:",
      df['Ticket_Price'].std())


# ============================================================
# 11. NUMPY - CREATE 2D RANDOM DATA
# ============================================================

print("\n" + "=" * 60)
print("NUMPY 2-DIMENSIONAL RANDOM DATA")
print("=" * 60)


random_data = np.random.randint(
    1,
    100,
    size=(5, 2)
)

print(random_data)


# ============================================================
# 12. NUMPY STATISTICAL OPERATIONS
# ============================================================

print("\n" + "=" * 60)
print("NUMPY STATISTICAL OPERATIONS")
print("=" * 60)


print("Mean:",
      np.mean(random_data))

print("Median:",
      np.median(random_data))

print("Standard Deviation:",
      np.std(random_data))

print("Maximum:",
      np.max(random_data))

print("Minimum:",
      np.min(random_data))


# ============================================================
# 13. SAVE CLEANED DATA
# ============================================================

df.to_csv(
    "cleaned_visitor_data.csv",
    index=False
)

print("\nCleaned dataset saved as: cleaned_visitor_data.csv")


# ============================================================
# 14. MATPLOTLIB - SINE WAVE
# ============================================================

x = np.linspace(
    0,
    2 * np.pi,
    200
)

amplitude = np.sin(x)


fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    x,
    amplitude,
    linewidth=2,
    label="Sine Wave"
)

ax.set_title(
    "Sine Wave",
    fontsize=14,
    fontweight='bold'
)

ax.set_xlabel(
    "Time / Angle",
    fontsize=12
)

ax.set_ylabel(
    "Amplitude",
    fontsize=12
)

ax.grid(
    True,
    linestyle='--',
    alpha=0.6
)

ax.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 15. ENGINEERING DOMAIN
#     DAMPED HARMONIC RESPONSE OF STRUCTURE
# ============================================================


# Time in seconds
time = np.linspace(
    0,
    10,
    1000
)


# Engineering parameters

initial_amplitude = 10       # mm
damping = 0.25               # damping coefficient
frequency = 2 * np.pi        # angular frequency


# Damped harmonic equation

amplitude = (
    initial_amplitude
    * np.exp(-damping * time)
    * np.sin(frequency * time)
)


# Create figure

fig, ax = plt.subplots(
    figsize=(10, 5)
)


# Plot response

ax.plot(
    time,
    amplitude,
    color='crimson',
    linewidth=2,
    label="Sensor A Displacement"
)


# Title

ax.set_title(
    'Damped Harmonic Response of Structure',
    fontsize=14,
    fontweight='bold',
    pad=15
)


# Axis labels

ax.set_xlabel(
    'Time (seconds)',
    fontsize=12
)

ax.set_ylabel(
    'Displacement (mm)',
    fontsize=12
)


# Enable engineering gridlines

ax.grid(
    True,
    linestyle='--',
    alpha=0.6
)


# Configure the Legend

ax.legend(
    loc='upper right',
    frameon=True,
    shadow=True
)


plt.tight_layout()

plt.show()


# ============================================================
# 16. FINAL CLEAN DATASET
# ============================================================

print("\n" + "=" * 60)
print("FINAL CLEANED DATASET")
print("=" * 60)

print(df)


print("\n" + "=" * 60)
print("ALL EXPERIMENTS COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
